```python
#16
a16=df.groupby("branch")["units_sold"].agg(["sum", "mean", "min", "max", "std"])
print(a16)
```

```python
#17
a17=df.groupby("product").agg(total_units_sold=("units_sold", "sum"),avgd=("discount_pct", "mean"),maxr=("customer_rating", "max"),mindt=("delivery_minutes", "min"))
print(a17)
```

```python
#18
print("Total")
print(df.groupby("branch").size())
print("\nNon-missing")
print(df.groupby("branch")["customer_rating"].count())
```

```python
#19
def slr(values):
    return values.max()-values.min()
a19=df.groupby("product")["units_sold"].agg(slr)
print(a19)
```

```python
#20
a20=df.groupby(["branch","product"],as_index=False).agg(tus=("units_sold","sum"),ttu=("target_units","sum"))
print(a20)
```

```python
#21
for branch, records in df.groupby("branch"):
    print("Branch:", branch)
    print(records)
    print("Total units sold:", records["units_sold"].sum())
    print()
```

```python
#22
a22=(df.groupby("product")["units_sold"].quantile([0.25, 0.50, 0.75]).unstack())
print(a22)
```

```python
#23
ind=df.set_index(["branch","product"])
a23=ind.groupby(level="branch")["units_sold"].sum()
print(a23)
```

```python
#24
a24=pd.pivot_table(df,values="units_sold",index="branch",columns="product",aggfunc="sum",margins=True,margins_name="Total")
print(a24)
```

```python
#25
a25=pd.pivot_table(df,values=["discount_pct", "customer_rating"],index="branch",columns="product",aggfunc="mean")
print(a25)
```

```python
#26
df["customer_rating"] = df["customer_rating"].fillna(df.groupby(["branch", "product"])["customer_rating"].transform("mean"))
print(df)
```

```python
#27
def dis(group):
  t=(group["units_sold"]*group["discount_pct"]).sum()
  tu=group["units_sold"].sum()
  return t/tu

a27=df.groupby("branch")[["units_sold","discount_pct"]].apply(dis)
print(a27)
```

```python
#28
sorted_df = df.sort_values("units_sold", ascending=False)
a28= sorted_df.groupby("branch").head(3)
print(a28[["branch", "product", "day", "units_sold"]])
```

```python
#29
branch_total = df.groupby("branch")["units_sold"].transform("sum")

df["sales_contribution_pct"] = df["units_sold"] / branch_total * 100

print(df[["branch", "units_sold", "sales_contribution_pct"]])
print(df.groupby("branch")["sales_contribution_pct"].sum().round(2))
```

```python
#30
df["sales_category"] = pd.qcut(
    df["units_sold"],
    4,
    labels=[
        "Low Sales",
        "Below Average Sales",
        "Above Average Sales",
        "High Sales"
    ]
)

print(
    df.groupby("sales_category", observed=False).agg({
        "units_sold": "sum",
        "discount_pct": "mean",
        "customer_rating": "mean",
        "delivery_minutes": "mean",
        "record_id": "size"
    })
)
```
