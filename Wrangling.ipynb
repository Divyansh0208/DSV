```python
#27
combined = orig.combine_first(upd)
r = pd.merge(students,combined[["Marks"]].reset_index(),on="Student_ID",how="left")
print(r)
print("\ncomparison")
comp = pd.concat([orig["Marks"].rename("Original"),upd["Marks"].rename("Updated"),combined["Marks"].rename("Final")],axis=1)
print(comp.loc[[101, 102]])
```

```python
#28
f=orig.combine_first(upd)
h=orig.combine(upd, np.fmax)
comp=pd.concat([f["Marks"].rename("combine_first"),h["Marks"].rename("fmax")],axis=1).sort_index()
same=(comp["combine_first"].eq(comp["fmax"]) | (comp["combine_first"].isna() & comp["fmax"].isna()))
different=comp[~same]
print(comp)
print("\nRows")
print(different)
print("\nStudent IDs")
print(different.index.tolist())
```
